```python

import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Input and target
x1, x2 = 0.5, 0.8
target = 1.0
lr = 0.5

# Initial weights and biases
w1, w2 = 0.15, 0.20
w3, w4 = 0.25, 0.30
v1, v2 = 0.40, 0.45
b1, b2, bo = 0.35, 0.35, 0.60

# FORWARD PASS 1
h1 = sigmoid(x1*w1 + x2*w2 + b1)
h2 = sigmoid(x1*w3 + x2*w4 + b2)
out = sigmoid(h1*v1 + h2*v2 + bo)

previous_error = 0.5 * (target - out)**2

print("FORWARD PASS 1")
print("Output         :", round(out, 6))
print("Previous Error :", round(previous_error, 6))

# BACKWARD PASS
delta_o = (target - out) * out * (1 - out)
delta_h1 = h1 * (1 - h1) * delta_o * v1
delta_h2 = h2 * (1 - h2) * delta_o * v2

# Update weights and biases
v1 += lr * delta_o * h1
v2 += lr * delta_o * h2
bo += lr * delta_o

w1 += lr * delta_h1 * x1
w2 += lr * delta_h1 * x2
b1 += lr * delta_h1

w3 += lr * delta_h2 * x1
w4 += lr * delta_h2 * x2
b2 += lr * delta_h2

# FORWARD PASS 2
h1 = sigmoid(x1*w1 + x2*w2 + b1)
h2 = sigmoid(x1*w3 + x2*w4 + b2)
out = sigmoid(h1*v1 + h2*v2 + bo)

new_error = 0.5 * (target - out)**2

# ERROR REDUCTION
error_reduction = previous_error - new_error

print("\nFORWARD PASS 2")
print("New Output     :", round(out, 6))
print("New Error      :", round(new_error, 6))

print("\nERROR CALCULATION")
print("Previous Error :", round(previous_error, 6))
print("New Error      :", round(new_error, 6))
print("Error Reduction:", round(error_reduction, 6))

print("\nForward Passes  : 2")
print("Backward Passes : 1")
```
